# 02 — Data Preprocessing

Notebook ini mendokumentasikan aturan pembersihan dataset mentah (`dataset/raw/*.csv`) menjadi dataset bersih (`dataset/processed/*.csv`) sesuai PRD:
1. **Sleep Health & Lifestyle:**
   - Ambil 4 fitur: `Sleep Duration`, `Quality of Sleep`, `Physical Activity Level`, `Stress Level`.
   - Isi `NaN` pada `Sleep Disorder` dengan `'None'` (jangan di-`dropna`).
   - Konversi target biner: `Insomnia` = `1`, lainnya = `0`.
2. **Student Productivity:**
   - Ambil 5 fitur: `sleep_hours`, `phone_usage_hours`, `coffee_intake_mg`, `exercise_minutes`, `stress_level`.
   - Binning `productivity_score` menjadi 3 kelas: `0` (`Rendah`), `1` (`Sedang`), `2` (`Tinggi`).

```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path("..") if Path("../dataset").exists() else Path(".")
RAW_SLEEP = BASE_DIR / "dataset" / "raw" / "sleep_health_lifestyle.csv"
RAW_PROD = BASE_DIR / "dataset" / "raw" / "student_performance.csv"
OUT_SLEEP = BASE_DIR / "dataset" / "processed" / "sleep_cleaned.csv"
OUT_PROD = BASE_DIR / "dataset" / "processed" / "productivity_cleaned.csv"

SLEEP_COLS = ["Sleep Duration", "Quality of Sleep", "Physical Activity Level", "Stress Level"]
PROD_COLS = ["sleep_hours", "phone_usage_hours", "coffee_intake_mg", "exercise_minutes", "stress_level"]
```

```python
if RAW_SLEEP.exists() and RAW_SLEEP.stat().st_size > 0:
    df_sleep = pd.read_csv(RAW_SLEEP)
    df_sleep["Sleep Disorder"] = df_sleep["Sleep Disorder"].fillna("None")
    df_sleep["target"] = (df_sleep["Sleep Disorder"] == "Insomnia").astype(int)
    df_sleep_clean = df_sleep[SLEEP_COLS + ["Sleep Disorder", "target"]].copy()
    df_sleep_clean.to_csv(OUT_SLEEP, index=False)
    print("Saved:", OUT_SLEEP, df_sleep_clean.shape)
else:
    df_sleep_clean = pd.read_csv(OUT_SLEEP)
    print("Raw CSV kosong/tidak ada; memuat hasil bersih:", OUT_SLEEP, df_sleep_clean.shape)

display(df_sleep_clean.head())
```

```python
if RAW_PROD.exists() and RAW_PROD.stat().st_size > 0:
    df_prod = pd.read_csv(RAW_PROD)
    df_prod_clean = df_prod[PROD_COLS + ["productivity_score"]].dropna().copy()
    df_prod_clean["productivity_class"] = pd.qcut(
        df_prod_clean["productivity_score"], q=3, labels=[0, 1, 2]
    ).astype(int)
    label_map = {0: "Rendah", 1: "Sedang", 2: "Tinggi"}
    df_prod_clean["class_name"] = df_prod_clean["productivity_class"].map(label_map)
    df_prod_clean.to_csv(OUT_PROD, index=False)
    print("Saved:", OUT_PROD, df_prod_clean.shape)
else:
    df_prod_clean = pd.read_csv(OUT_PROD)
    print("Raw CSV kosong/tidak ada; memuat hasil bersih:", OUT_PROD, df_prod_clean.shape)

display(df_prod_clean.head())
```
